```python
from pyspark.sql.functions import when, col
```

```python
df_silver = spark.table("project_cardio.silver.cardio_clean")
```

```python

df_gold_ml = (
    df_silver
    .withColumn(
        "bp_category",
        when(col("tension_systolic") >= 140, "high")
        .when(col("tension_systolic") >= 120, "elevated")
        .otherwise("normal")
    )
    .withColumn(
        "bmi_category",
        when(col("bmi") < 18.5, "underweight")
        .when((col("bmi") >= 18.5) & (col("bmi") < 25), "normal")
        .when((col("bmi") >= 25) & (col("bmi") < 30), "overweight")
        .otherwise("obese")
    )
    .select(
        "id",
        "age_years",
        "gender",
        "height",
        "weight",
        "bmi",
        "bmi_category",
        "tension_systolic",
        "tension_diastolic",
        "bp_category",
        "cholesterol",
        "gluc",
        "smoke",
        "alco",
        "active",
        "cardio"
    )
)
```

```python
df_gold_ml.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("project_cardio.gold.cardio_features")
```
