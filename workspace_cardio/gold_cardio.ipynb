```python
from pyspark.sql.functions import avg, count, sum
```

```python
df_silver = spark.table("project_cardio.silver.cardio_clean")
```

```python
df_gold_dashboard = (
    df_silver 
    .groupBy("gender")
    .agg(
        count("*").alias("nb_patients"),
        avg("age_years").alias("avg_age"),
        avg("bmi").alias("avg_bmi"),
        avg("tension_systolic").alias("avg_tension_systolic"),
        avg("tension_diastolic").alias("avg_tension_diastolic"),
        avg("cholesterol").alias("avg_cholesterol"),
        avg("gluc").alias("avg_gluc"),
        sum("cardio").alias("nb_cardio_risk")
    )
)
```

```python
df_gold_dashboard.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("project_cardio.gold.cardio_dashboard_metrics")
```
