```python
from pyspark.sql.functions import current_timestamp, lit
```

```python
df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Volumes/project_cardio/bronze/volum_card/cardio.csv")
```

```python
df_bronze = df.withColumn("source_file", lit("cardio.csv")) \
              .withColumn("ingestion_timestamp", current_timestamp()) \
              .withColumn("source_type", lit("csv"))
```

```python
display(df_bronze)
```

```python
df_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("project_cardio.bronze.cardio_csv")
```
